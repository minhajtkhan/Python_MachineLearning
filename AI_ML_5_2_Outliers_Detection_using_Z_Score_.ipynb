The Z-Score Method is one of the most common statistical techniques for detecting outliers. It measures how far a data point is from the mean in terms of standard deviations.

A Z-Score tells us:

How many standard deviations a value is away from the mean.

Think of it like this:

Z = 0 → Exactly at the mean

Z = 1 → One standard deviation above the mean
Z = -2 → Two standard deviations below the mean

The farther the Z-score is from zero, the more unusual the observation.

The Formula
Z=
X−μ
/
σ


Where:

X = Data value

μ = Mean

σ = Standard Deviation

**How Do We Detect Outliers?**

| Z-Score                         | Interpretation |
| ------------------------------- | -------------- |
| Between -2 and +2               | Normal         |
| Between -3 and -2 or +2 and +3  | Unusual        |
| Less than -3 or Greater than +3 | Outlier        |


```python
import pandas as pd

df = pd.read_csv("used_cars_data.csv")
df.head()
```

Calculating Z-Score using Scipy

```python
from scipy import stats

# Calculate Z-score for 'Kilometers_Driven' and store in a new column
df["KM_Zscore"] = stats.zscore(df["Kilometers_Driven"])

# Calculate Z-score for 'Price' as well
df["Price_Zscore"] = stats.zscore(df["Price"])

# Display the dataset with the new Z-score columns
df[["Name", "Kilometers_Driven", "KM_Zscore", "Price", "Price_Zscore"]].head()
```

I got NaN because there are some Missing Values in the Price Column so find out how many values there are Missing we used this Code

```python
print(df["Price"].isna().sum())
```

There are 1234 Rows which have Missing Values

**Using Pure Pandas Formulas**

If you prefer doing it manually with pandas operations, the mathematical formula is:

```python
# Formula: (x - mean) / std_dev
df["KM_Zscore"] = (
    df["Kilometers_Driven"] - df["Kilometers_Driven"].mean()
) / df["Kilometers_Driven"].std()

print(df["KM_Zscore"])
```

**Find Outliers in Your Dataset**

you have the Z-scores, you can filter for values where the Z-score is greater than $3$ or less than $-3$:

```python
# Filter rows where Kilometers_Driven is an outlier (|Z-score| > 3)
outliers = df[df["KM_Zscore"].abs() > 3]

print("Outliers found:")
outliers[["Name", "Kilometers_Driven", "KM_Zscore"]]
```
