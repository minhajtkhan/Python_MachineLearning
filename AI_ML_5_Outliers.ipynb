**Why Are Outliers Important?**

Outliers can:

Affect Average (Mean)
Reduce Model Accuracy
Mislead Analysis
Create Wrong Predictions

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("used_cars_data.csv")

print(df.columns)

plt.figure(figsize=(6,4))

sns.boxplot(x=df['Kilometers_Driven'])

plt.title("Boxplot of Kilometers")

plt.show()
```

This Method of detection is called BoxPlot but even still we don't get an exact count of Outliers present for that we use Interquartile Range

The IQR (Interquartile Range) method is one of the most common ways to detect outliers in a dataset. It is simple, reliable, and works well even when the data is not normally distributed.

Step 1: What are Quartiles?

Imagine you have the following sorted dataset:

| Data |
| ---: |
|    5 |
|    7 |
|    8 |
|   10 |
|   12 |
|   15 |
|   18 |
|   20 |
|   25 |


Quartiles divide the data into 4 equal parts.

Q1           Q2            Q3
|------------|-------------|
25%         50%          75%

There are three important quartiles:

Q1 (First Quartile) → 25% of the data lies below this value.

Q2 (Second Quartile) → The median (50%).

Q3 (Third Quartile) → 75% of the data lies below this value.
