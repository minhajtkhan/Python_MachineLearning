**Theory**

Every supervised machine learning problem starts with splitting a dataset into two conceptual pieces:

Features (X) — the input variables the model is allowed to look at to make a prediction. Also called independent variables, predictors, or inputs.
Target (y) — the single variable the model is trying to predict. Also called the dependent variable, label, or output.

The relationship the model learns is essentially:

y = f(X) + error

*Where f is the pattern the algorithm tries to approximate from data.*

**Why the Distinction Matters**

The model must never be given the target as part of its input — that would be like giving a student the answer key mixed in with the exam questions.

Choosing the right target defines what kind of problem you're solving: continuous target → regression; categorical target → classification.

Everything downstream — train/test splitting, preprocessing, evaluation metrics — is built around this X/y separation.

Our dataset has 15 usable columns (after dropping Car_ID as discussed in the earlier lecture):

Brand, Model, Year, Age_Years, Price, Kilometer, Fuel_Type, Transmission,
Owner_Type, Mileage_kmpl, Engine_CC, Power_BHP, Seats, Color, City

Our business question: predict the resale price of a used car.

That makes the decision straightforward:

y (target): Price — a continuous numeric value, so this is a regression problem.
X (features): every other column — Brand, Model, Year/Age_Years, Kilometer, Fuel_Type, Transmission, Owner_Type, Mileage_kmpl, Engine_CC, Power_BHP, Seats, Color, City.

```python
import pandas as pd

df = pd.read_csv("used_cars_dataset.csv")
df = df.drop(columns=["Car_ID"])   # pure identifier, not a feature (see earlier lecture)

# Define target
y = df["Price"]

# Define features — everything except the target
X = df.drop(columns=["Price"])

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nFeature columns:")
print(list(X.columns))

print("\nTarget summary:")
print(y.describe())
```

Sanity Checks Worth Running

```python
# Confirm the target isn't accidentally still in X
assert "Price" not in X.columns

# Confirm no row mismatch between X and y
assert len(X) == len(y)

# Quick look at target distribution — is it skewed? (common for price data)
import matplotlib.pyplot as plt
y.hist(bins=40)
plt.title("Distribution of Price")
plt.xlabel("Price")
plt.show()
```

*Note: Used-car prices are frequently right-skewed (many affordable cars, a long tail of expensive ones). This observation will matter later when choosing evaluation metrics and possibly applying a log-transform to Price.*
