Once we have X and y, we need to split the data into separate groups so we can train a model on some data and honestly evaluate it on data it has never seen.

If we train and evaluate on the exact same rows, the model can simply memorize the answers — that tells us nothing about how it will perform on a car it hasn't seen before. This is the core reason the train-test split exists.

**The Basic Idea**

Training set — the rows the model learns from (fits its parameters on).
Test set — the rows held back, used only at the very end to measure real-world performance.

A common split ratio is 80% train / 20% test, though this varies with dataset size (smaller datasets sometimes use 70/30 or 90/10 depending on how much data can be spared for testing).

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("used_cars_dataset.csv").drop(columns=["Car_ID"])

X = df.drop(columns=["Price"])
y = df["Price"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% held out for testing
    random_state=42      # reproducibility
)

print("Training set size:", X_train.shape)
print("Test set size:", X_test.shape)

# Sanity check: no overlap between train and test rows
assert set(X_train.index).isdisjoint(set(X_test.index))
```
