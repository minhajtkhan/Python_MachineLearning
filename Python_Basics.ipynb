**Exception Handling in Python**

An exception is an error that occurs while a program is running and interrupts its normal execution.

```python
num1 = 10
num2 = 0

result = num1 / num2

print(result)
print("Program finished")
```

Python raises a ZeroDivisionError because division by zero is not allowed. The program does not reach the final print statement.

In real-world applications, unexpected input, missing files, or invalid calculations can cause similar problems.

**The try and except Blocks**

```python
try:
    # Code that might cause an error
    pass

except SomeException:
    # Handle the error
    pass
```

```python
try:
    num1 = 10
    num2 = 0

    result = num1 / num2
    print(result)

except ZeroDivisionError:
    print("You cannot divide by zero.")

print("Program finished")
```

**Handling Multiple Exceptions**

```python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2
    print(f"Result: {result}")

except ValueError:
    print("Please enter valid whole numbers.")

except ZeroDivisionError:
    print("The second number cannot be zero.")
```

**The finally Block**

The finally block runs whether an exception occurs or not, provided normal Python control flow reaches it.

```python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Invalid input.")

finally:
    print("Program execution completed.")
```
