**Outlier Detection using K-Nearest Neighbors (KNN)**

While methods like IQR analyze a single column at a time, KNN allows us to evaluate multiple features simultaneously (like comparing Price and Kilometers_Driven together).

The intuition behind KNN outlier detection is simple:Normal points live in dense neighborhoods with nearby friends. Outliers live in isolated regions far away from everyone else.

How It Works Step-by-StepPick $K$ (Number of Neighbors):

You choose how many neighbors each point should look at (e.g., $k = 5$).

Calculate Distances: For every data point in your dataset, compute its straight-line distance (Euclidean distance) to every other point.

Find $K$ Nearest Points: For each point, sort those distances and keep the $k$-closest ones.

Compute the Outlier Score: Take the average of those $k$ distances (or the distance to the $k$-th neighbor).

Flag Outliers: Points with unusually large average distances are far away from the rest of the dataset $\rightarrow$ Outliers.

Why Feature Scaling is EssentialDistance formulas use spatial math like Euclidean distance:$$\text{Distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$
If Kilometers_Driven ranges from $10,000$ to $200,000$ and Price ranges from $5$ to $50$, the huge numbers in Kilometers_Driven will dominate the calculation.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors

# 1. Load data and select numerical features
df = pd.read_csv("used_cars_data.csv")
X = df[["Kilometers_Driven", "Price"]].dropna()

# 2. Scale features so distances are equal
X_scaled = StandardScaler().fit_transform(X)

# 3. Fit KNN with 5 neighbors
knn = NearestNeighbors(n_neighbors=5).fit(X_scaled)

# 4. Compute outlier score (average distance to 5 nearest neighbors)
distances, _ = knn.kneighbors(X_scaled)
X["outlier_score"] = distances.mean(axis=1)

# 5. Flag top 5% highest scores as outliers
threshold = X["outlier_score"].quantile(0.95)
outliers = X[X["outlier_score"] > threshold]

print(outliers)
```
