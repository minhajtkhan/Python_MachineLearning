Imagine lining up all your data points from smallest to largest and splitting them into four equal quarters:

Q1 (First Quartile): The 25th percentile—25% of the data falls below this.

Q2 (Median): The 50th percentile—the middle point.  

Q3 (Third Quartile): The 75th percentile—75% of the data falls below this.  The Interquartile Range is simply the distance between Q3 and Q1:  
$$\text{IQR} = Q_3 - Q_1$$

To flag outliers, we set up "fences" (cut-off boundaries):  

Lower Boundary: $Q_1 - (1.5 \times \text{IQR})$

Upper Boundary: $Q_3 + (1.5 \times \text{IQR})$

Any data point below the Lower Boundary or above the Upper Boundary is classified as an outlier.

https://www.researchgate.net/publication/339459172/figure/fig3/AS:941482634641432@1601478360515/Graphical-display-of-IQR-method.png

```python
import pandas as pd

df = pd.read_csv("used_cars_data.csv")

# 1. Calculate Q1, Q3, and IQR for a column
Q1 = df["Kilometers_Driven"].quantile(0.25)
Q3 = df["Kilometers_Driven"].quantile(0.75)
IQR = Q3 - Q1

# 2. Define upper and lower bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 3. Filter outliers
outliers = df[(df["Kilometers_Driven"] < lower_bound) | (df["Kilometers_Driven"] > upper_bound)]

print(outliers)
```
