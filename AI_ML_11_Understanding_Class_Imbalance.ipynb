# What Is Class Imbalance?

Class imbalance happens when the categories (classes) you're trying to predict in a classification problem are not represented equally in your dataset. One class — the majority class — has far more examples than the other(s), the minority class(es).

This isn't a rare edge case. It's one of the most common real-world problems in machine learning:

Fraud detection: 99.9% legitimate transactions, 0.1% fraud
Disease diagnosis: most patients are healthy, few have the rare disease
Spam detection: most emails are not spam
Our dataset: most used cars run on Petrol, very few are Hybrid

# Why Class Imbalance Is a Problem

Imagine we train a classifier on this data to predict Fuel_Type. A "lazy" model could simply predict "Petrol" every single time and still be right 77.7% of the time. That sounds like good accuracy — but the model has learned nothing about what makes a car a Diesel, CNG, or Hybrid vehicle. It is functionally useless for identifying the minority classes, which are often the ones we care about most (e.g., flagging fraud, catching disease, or here, correctly identifying rarer eco-friendly Hybrid vehicles).

# Better Evaluation Metrics

Because of this, when working with imbalanced data we favor metrics that look at each class individually:

1. Precision — of everything predicted as the minority class, how much was correct?

2. Recall (Sensitivity) — of all actual minority-class examples, how many did we catch?

3. F1-score — the harmonic mean of precision and recall

4. Confusion Matrix — a full breakdown of correct/incorrect predictions per class

# Why Does Class Imbalance Bias a Model?

Most machine learning algorithms (logistic regression, decision trees, neural networks, etc.) are trained by minimizing an overall error/loss across all training examples. If 77.7% of your training rows say "Petrol," the optimization process is naturally pulled toward minimizing error on Petrol cars — because that's where most of the loss comes from. The minority classes contribute so little to the total loss that the model can largely ignore them and still score well.
