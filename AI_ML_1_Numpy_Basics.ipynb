**INDEXING SLICING & RESHAPING**

1. Slicing in Multi-Dimensions

    NumPy uses [row_slice, column_slice] notation.

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Extract single element (row index 1, col index 2) -> 60
print(matrix[1, 2])

# Extract first two rows, last two columns
# Output: [[20, 30], [50, 60]]
print(matrix[0:3, 1:])
```

```python
# Extract the second row (index 1) across ALL columns
print("Row 1:", matrix[1, :])
# Output: [40 50 60]

# Extract the third column (index 2) across ALL rows
print("Column 2:\n", matrix[:, 2])
# Output: [30 60 90]
```

```python
# Rows 1 to end, Columns 0 to 2 (excludes col index 2)
print("Bottom-left 2x2:\n", matrix[1:, 0:2])
# Output:
# [[40 50]
#  [70 80]]
```

2. Reshaping Arrays

    Change dimensions without altering the underlying data using .reshape().

```python
arr = np.arange(1, 7)  # [1, 2, 3, 4, 5, 6]

# Convert 1D (6 elements) -> 2D (2 rows, 3 columns)
matrix_2x3 = arr.reshape(2, 3)

# Use -1 to let NumPy calculate the dimension automatically
matrix_3x2 = arr.reshape(3, -1)  # Infers 2 columns automatically
```

**Vectorization & Arithmetic Operations**

Vectorization allows operating on entire arrays at once without needing explicit for loops.

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b)   # [11, 22, 33]
print(a * b)   # [10, 40, 90]
print(a ** 2)  # [1, 4, 9]
```

Summary Statistics (Aggregations)

```python
data = np.array([[1, 2], [3, 4]])

np.sum(data)         # 10 (sum of all elements)
np.sum(data, axis=0) # [4, 6] (sum down the columns)
np.sum(data, axis=1) # [3, 7] (sum across the rows)

np.mean(data)        # Average
np.std(data)         # Standard deviation
```

The mean and standard deviation tell you two core things about your data:Mean: Where the "center" or average of your data sits.Standard Deviation: How spread out the data points are around that center.1. The Mean (Average)The mean is the average value. You calculate it by adding up all the numbers and dividing by how many numbers there are.Formula:$$\text{Mean } (\mu) = \frac{\sum x}{N}$$Real-World ExampleImagine three test scores: 80, 85, and 90.$$\text{Mean} = \frac{80 + 85 + 90}{3} = \frac{255}{3} = 85$$2. Standard Deviation (The Spread)Standard deviation measures how far, on average, data points drift away from the mean.Low Standard Deviation: The numbers are tightly packed around the mean.High Standard Deviation: The numbers are spread out over a wide range.Why do we need it?Consider two classes that both have an average test score of 80:Class A scores: 79, 80, 81 $\rightarrow$ Low standard deviation (very consistent).Class B scores: 50, 80, 110 $\rightarrow$ High standard deviation (wildly spread out).Just knowing the mean (80) hides how different these two classes actually are!How Standard Deviation is Calculated (Step-by-Step)The formula looks like this:$$\sigma = \sqrt{\frac{\sum (x - \mu)^2}{N}}$$Here is how you break it down into simple steps using the scores [70, 80, 90]:Find the Mean:$$\mu = \frac{70 + 80 + 90}{3} = 80$$Subtract the mean from each value (Find deviations):$70 - 80 = -10$$80 - 80 = 0$$90 - 80 = 10$Square each difference (this eliminates negative signs):$(-10)^2 = 100$$(0)^2 = 0$$(10)^2 = 100$Find the mean of those squared differences (Variance):$$\text{Variance} = \frac{100 + 0 + 100}{3} = \frac{200}{3} \approx 66.67$$Take the square root (this converts the units back to normal):$$\text{Standard Deviation } (\sigma) = \sqrt{66.67} \approx 8.16$$

```python
import numpy as np

scores = np.array([70, 80, 90])

# Calculate Mean
mean_val = np.mean(scores)
print("Mean:", mean_val)  # Output: 80.0

# Calculate Standard Deviation
std_val = np.std(scores)
print("Standard Deviation:", std_val)  # Output: 8.1649...
```
