About Our Dataset

Throughout this lecture, we'll use a Used Cars Dataset, similar to the popular Kaggle "Used Car Price Prediction" datasets. Here is what it typically looks like:

We'll load this in Python throughout the lecture:

```python
import pandas as pd

# Load the dataset (downloaded from Kaggle)
df = pd.read_csv('used_cars_dataset.csv')

# Preview first 5 rows
print(df.head())

# Check shape (rows, columns)
print("Shape:", df.shape)
```

**Step 1: Inspecting Column Data Types**

Before deciding which columns are numerical vs categorical, we always check how Python/Pandas has interpreted each column:

```python
print(df.dtypes)
```

**How to read this:**

int64 / float64 → Pandas has detected these as numbers → likely Numerical Data

object → Pandas has detected these as text/strings → likely Categorical Data

**Numerical Columns in Our Dataset**

**3.1 Discrete Numerical Columns**

These are countable whole numbers:

```python
discrete_cols = ['year', 'num_seats']
print(df[discrete_cols].describe())
```

**3.2 Continuous Numerical Columns**

These can take any value in a range, including decimals:

```python
continuous_cols = ['Kms_Driven', 'Mileage_kmpl', 'Engine_CC', 'Power_bhp', 'Price_Lakh']
print(df[continuous_cols].describe())
```

# **Categorical Columns in Our Dataset**

**4.1 Nominal Categorical Columns (No Order)**

```python
nominal_cols = ['Car_Name', 'Brand', 'Fuel_Type', 'Seller_Type', 'Transmission']
for col in nominal_cols:
    print(f"\n{col} unique values:", df[col].unique())
```

**4.2 Ordinal Categorical Columns (Has Order)**

```python
print(df['Owner'].unique())
# Output: ['First' 'Second' 'Third' 'Fourth & Above']
```

# Visualizing the Difference (Practical Teaching Aid)

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Numerical -> Histogram
axes[0].hist(df['Price_Lakh'], bins=10, color='steelblue', edgecolor='black')
axes[0].set_title('Price_Lakh (Numerical - Continuous)')
axes[0].set_xlabel('Price (Lakh PKR/INR)')

# Categorical -> Bar Chart
df['Fuel_Type'].value_counts().plot(kind='bar', ax=axes[1], color='darkorange')
axes[1].set_title('Fuel_Type (Categorical - Nominal)')

plt.tight_layout()
plt.show()
```

**Explanation:**

Numerical data → Histogram (shows distribution/spread of continuous values)
Categorical data → Bar Chart (shows count/frequency of each category)
value_counts() counts how many rows fall into each category — the standard first step for exploring categorical columns.

Teaching Point: If a bar chart looks meaningful for a column, and a histogram looks meaningless (or vice versa), that's a strong practical hint about the column's true type.
